```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
rng=np.random.RandomState(42) # generate a random value and seed that
x= 10*rng.rand(50) #generate 50 random value
y=2*x-1+rng.randn(50) #generate 50 random noise value
plt.scatter(x,y);
```

```python
#Choose a class of model by importing the appropriate estimator class from scikit learn
from sklearn.linear_model import LinearRegression as
```

```python
#Choose model hyperparameters by instantiating this class with desired values
model = LinearRegression(fit_intercept=True)
model
```

```python
#Arrange data into a features matrix and target vector following the discussion from before
X=x[:,np.newaxis]
X.shape
```

```python
#Fit the model to your data by calling the fit() method of the model instance
model.fit(X,y)
```

```python
model.coef_ #y= mx+c coef refers the value of m
```

```python
model.intercept_ # y= mx+c Interception refers the value of c
```

```python
# Apply the model to new data
xfit=np.linspace(-1,11)
```

```python
Xfit=xfit[:,np.newaxis]
yfit=model.predict(Xfit)
```

```python
plt.scatter(x,y)
plt.plot(xfit,yfit, color="red");
```

```python
#implementation of linear regression
df = pd.read_csv("https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day48-simple-linear-regression/placement.csv")
df.shape
df.head()
```

```python
plt.scatter(df["cgpa"],df["package"])
plt.xlabel("CGPA" , color="red")
plt.ylabel("Package",color="red")
plt.show()
```

```python
X=df.iloc[:,0:1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2) #here x is feature data and y is target data .here spilt method devide the data for train and test here test size .2 that means 80% data for train and 20% data for testing and that devidation done via randomly. here used seed and random_state=2 .
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train) # here training the data by using fit method
```

```python
X_test #checking the test case for feature data
```

```python
y_test # checking the test case for targeted data
```

```python
lr.predict(X_test.iloc[3].values.reshape(1,1)) #now predicting data for new value
# here data frame values converted in 1 X 1 numpy array and then predict the value of y for X_test
```

```python
#Visualizations of liner regration prediction
plt.scatter(df["cgpa"],df["package"])
plt.plot(X_train,lr.predict(X_train),color='red')
plt.xlabel("CGPA" , color="red")
plt.ylabel("Package(in lpa)",color="red")
plt.show()
```
